# Media de una o dos poblaciones normales con varianzas desconocidas: distribución t de Student

Desarrollo de los ejercicios de inferencia sobre medias cuando la varianza poblacional no se
conoce. Para una sola población se usa la t de una muestra; para dos poblaciones independientes
con varianzas iguales, la t combinada (pooled) con los grados de libertad sumados. Cada sección
presenta los datos, el estadístico y su distribución bajo la nula, la región de rechazo, el valor
p y el intervalo correspondiente, con el contraste de dualidad entre intervalo y prueba, la
verificación previa del supuesto de varianzas iguales cuando aplica, y una simulación Monte Carlo
con semilla fija.

## Marco teórico

### Distribución t de Student

Si $Z \sim N(0,1)$ y $U \sim \chi^2_\nu$ son independientes, entonces

$$T = \frac{Z}{\sqrt{U/\nu}} \ \sim\ t_\nu$$

con $\nu$ grados de libertad. La distribución es simétrica respecto de 0, con media $E[T] = 0$
para $\nu > 1$ y varianza $\operatorname{Var}(T) = \nu/(\nu - 2)$ para $\nu > 2$; no existe momento
de orden superior o igual a $\nu$. Comparada con la normal estándar tiene colas más pesadas, y esa
es justamente su función: absorbe la incertidumbre adicional de tener que estimar $\sigma$ con una
muestra finita. Con $\nu$ creciente converge a la normal. El aparecimiento de la t es inmediato:
en $T = Z/\sqrt{U/\nu}$ el numerador es normal estandarizado y el denominador es la raíz de una
chi-cuadrado sobre sus grados de libertad, que es exactamente lo que produce el error estándar
cuando $\sigma$ se reemplaza por $S$.

### Una sola población: intervalo y prueba para la media

Sea $X_1, \dots, X_n$ una muestra de $N(\mu, \sigma^2)$ con $\sigma^2$ desconocida. Entonces

$$T = \frac{\bar X - \mu}{S/\sqrt{n}} \ \sim\ t_{n-1} \qquad \text{(exacto, bajo normalidad)}$$

Por linealidad en el numerador y porque $S$ es el estimador de $\sigma$ con $n-1$ grados de
libertad. El intervalo de confianza resulta despejando $\mu$:

$$\bar x \ \pm \ t_{1-\alpha/2,\,n-1} \frac{s}{\sqrt{n}}$$

La prueba usa el mismo estadístico: $H_0: \mu = \mu_0$ contra $H_1: \mu \ne \mu_0$ (o
$< \mu_0$, $> \mu_0$), rechazando si $|t| > t_{1-\alpha/2,\,n-1}$ en el bilateral o si
$t < -t_{1-\alpha,\,n-1}$ en la cola izquierda. El valor p es la acumulada de $t_{n-1}$ hasta el
estadístico observado, o su complemento, según la cola.

### Dos poblaciones con varianzas iguales y desconocidas: t combinada

Sean $X_1, \dots, X_{n_1}$ e $Y_1, \dots, Y_{n_2}$ muestras independientes de
$N(\mu_1, \sigma^2)$ y $N(\mu_2, \sigma^2)$: misma varianza desconocida en ambas. La varianza
combinada (pooled) promedia las dos varianzas muestrales ponderadas por sus grados de libertad:

$$s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}$$

$$T = \frac{(\bar X - \bar Y) - \delta_0}{s_p\sqrt{1/n_1 + 1/n_2}} \ \sim\ t_{n_1 + n_2 - 2}
\ \text{bajo } H_0: \mu_1 - \mu_2 = \delta_0$$

$$\text{IC}_{1-\alpha}: \quad (\bar x - \bar y) \ \pm \ t_{1-\alpha/2,\,n_1+n_2-2} \
s_p\sqrt{\frac{1}{n_1} + \frac{1}{n_2}}$$

La lógica es la misma que con chi-cuadrado y F: bajo la nula existe una única varianza, y la
forma más eficiente de estimarla es combinar la información de los dos grupos. Con tamaños
iguales, la t combinada es robusta frente a varianzas desiguales; con tamaños desiguales en
proporción apreciable a las varianzas, el nivel real de la prueba se desvía del declarado y
conviene la aproximación de Welch.

### Verificación previa del supuesto de varianzas iguales

Antes de la t combinada puede testearse la igualdad de varianzas con la prueba F bilateral
ya vista: $H_0: \sigma_1^2 = \sigma_2^2$, estadístico $F = s_1^2/s_2^2$ y regiones en
ambas colas. Es un pre-test, y como tal conviene leerlo con reserva: si no se rechaza, la t
combinada es la herramienta adecuada; si se rechaza, el procedimiento combinado no es válido tal
cual. Además, condicionar la segunda prueba al resultado de la primera modifica un poco el nivel
real global, efecto despreciable cuando los tamaños son iguales.

### Dualidad y supuestos

La correspondencia intervalo-prueba sigue operando en las dos direcciones: un intervalo bilateral
de confianza $1-\alpha$ no contiene a $\mu_0$ si y sólo si la prueba bilateral de nivel $\alpha$
lo rechaza, y el límite de un intervalo unilatero de confianza $1-\alpha$ coincide con el umbral
de la prueba de cola simple de nivel $\alpha$. Nótese que eso implica que la cola simple al 5%
usa el cuantil de cola 0,05, no el de 0,025 del intervalo bilateral del 95%: son niveles de
intervalo distintos y las conclusiones pueden diferir. Supuestos: normalidad de las poblaciones
(indispensable con muestras chicas, donde no hay teorema central del límite que socorre),
independencia de las observaciones y de las dos muestras, muestreo aleatorio, y para la t
combinada además la igualdad de varianzas. Con $n$ grandes, $t_\nu$ se acerca a la normal y ambas
pruebas convergen a las versiones de varianza conocida.

### Cuantiles de referencia

En todas las fórmulas, $t_{q,\nu}$ y $F_{q,\nu_1,\nu_2}$ denotan el cuantil de probabilidad
acumulada $q$. Los usados en el desarrollo:

| Uso | Cuantil | Valor |
|---|---|---|
| IC 95% de $\mu$, $\nu = 11$ | $t_{0.975,\,11}$ | 2.200985 |
| Prueba unilateral 5% (cola izquierda), $\nu = 11$ | $t_{0.95,\,11}$ | 1.795885 |
| IC 90% de la diferencia, $\nu = 30$ | $t_{0.95,\,30}$ | 1.697261 |
| Prueba unilateral 1% (cola derecha), $\nu = 30$ | $t_{0.99,\,30}$ | 2.457262 |
| IC 95% de la diferencia (control), $\nu = 30$ | $t_{0.975,\,30}$ | 2.042272 |
| Prueba unilateral 5% (cola derecha), $\nu = 22$ | $t_{0.95,\,22}$ | 1.717144 |
| IC 95% de la diferencia, $\nu = 22$ | $t_{0.975,\,22}$ | 2.073873 |
| Prueba F bilateral 10%, $\nu = (15, 15)$, cola inferior | $F_{0.05,\,15,\,15}$ | 0.416069 |
| Prueba F bilateral 10%, $\nu = (15, 15)$, cola superior | $F_{0.95,\,15,\,15}$ | 2.403447 |


```python

import math

import numpy as np
import scipy
import matplotlib as mpl
import matplotlib.pyplot as plt
from scipy import stats

mpl.rcParams.update({
    "figure.figsize": (9, 4.5),
    "figure.dpi": 110,
    "axes.grid": True,
    "grid.alpha": 0.3,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
})

AZUL = "#1f4e79"
GRIS = "#808080"
VERDE = "#2e7d32"
ROJO = "#b03a2e"

t_dist = stats.t
f_dist = stats.f
norm_dist = stats.norm

# cuantiles de probabilidad acumulada q usados en el desarrollo
Q = {
    "t_0.975_11": 2.200985,
    "t_0.95_11": 1.795885,
    "t_0.95_30": 1.697261,
    "t_0.975_30": 2.042272,
    "t_0.99_30": 2.457262,
    "t_0.95_22": 1.717144,
    "t_0.975_22": 2.073873,
    "F_0.05_15_15": 0.416069,
    "F_0.95_15_15": 2.403447,
}

CHEQUEOS = [
    ("t", 0.975, 11),
    ("t", 0.95, 11),
    ("t", 0.95, 30),
    ("t", 0.99, 30),
    ("t", 0.975, 30),
    ("t", 0.95, 22),
    ("t", 0.975, 22),
    ("F", 0.05, (15, 15)),
    ("F", 0.95, (15, 15)),
]


def cuantil(dist, q, gl):
    return t_dist.ppf(q, gl) if dist == "t" else f_dist.ppf(q, *gl)


print(f"numpy {np.__version__} | scipy {scipy.__version__} | matplotlib {mpl.__version__}")
print()
print(f"{'distribucion':>14}{'q':>8}{'gl':>8}{'tabulado':>13}{'scipy':>13}{'|diff|':>10}")
print("-" * 66)
peor_prob = 0.0
for dist, q, gl in CHEQUEOS:
    scipy_q = cuantil(dist, q, gl)
    if dist == "t":
        tabulado, gl_txt = Q[f"t_{q}_{gl}"], f"{gl}"
        prob = t_dist.cdf(tabulado, gl)
    else:
        tabulado, gl_txt = Q[f"F_{q}_{gl[0]}_{gl[1]}"], f"{gl[0]},{gl[1]}"
        prob = f_dist.cdf(tabulado, *gl)
    peor_prob = max(peor_prob, abs(prob - q))
    nombre = "t-Student" if dist == "t" else "F"
    print(f"{nombre:>14}{q:>8}{gl_txt:>8}{tabulado:>13.6f}{scipy_q:>13.6f}"
          f"{abs(tabulado - scipy_q):>10.1e}")

print()
simetria = abs(t_dist.ppf(0.95, 30) + t_dist.ppf(0.05, 30))
reciproco = 1.0 / f_dist.ppf(0.95, 15, 15)
print(f"simetria de t: |t(0.95, 30) + t(0.05, 30)| = {simetria:.1e}   (t(0.05,30) = {-t_dist.ppf(0.95, 30):.6f})")
print(f"reciprocidad F: 1/F(0.95, 15, 15) = {reciproco:.6f}   vs   F(0.05, 15, 15) = {Q['F_0.05_15_15']:.6f}")
print(f"probabilidad acumulada sobre el tabulado: peor |F(cuantil) - q| = {peor_prob:.1e}")
```

## 1. Consumo anual de una aspiradora: intervalo del 95% y prueba unilateral

### Datos

Número anual de kilowatts-hora ($kWh$) consumidos por una aspiradora de marca ZZ, en hogares
tipo que usan ese modelo. Se supone que la variable poblacional es normal. Muestra aleatoria de
$n = 12$ hogares con promedio $\bar x = 42$ $kWh$/año y desviación estándar muestral
$s = 11,9$ $kWh$/año.

| Magnitud | Valor |
|---|---|
| Tamaño muestral, $n$ | 12 |
| Promedio muestral, $\bar x$ | 42,0 $kWh$/año |
| Desviación estándar muestral, $s$ | 11,9 $kWh$/año |
| Grados de libertad, $n - 1$ | 11 |

### a) Intervalo de confianza del 95% para el consumo promedio

Con $\\sigma$ desconocida se usa $T = (\\bar X - \\mu)/(S/\\sqrt{n}) \\sim t_{n-1}$:

$$\\bar x \\ \\pm \\ t_{0.975,\\,11} \\frac{s}{\\sqrt{n}} = 42,0 \\ \\pm \\ 2,200985 \\times
\\frac{11,9}{\\sqrt{12}} = 42,0 \\ \\pm \\ 7,56 = [\\,34{,}44;\\ 49{,}56\\,]$$

El consumo promedio anual de la marca ZZ se encuentra entre 34,44 y 49,56 $kWh$/año con un 95%
de confianza. El margen es bastante ancho: 7,56 $kWh$ sobre un promedio de 42, es cerca del 18%
del valor estimado, consecuencia directa de tener solo 12 observaciones y una variabilidad
poblacional grande respecto de la media ($s/\\bar x \\approx 0,28$).

### b) Prueba unilateral al 5% contra $\\mu_0 = 49$

Se quiere saber si hay evidencia de que el consumo promedio es **menor** a 49 $kWh$/año:

$$H_0: \\mu = 49 \\qquad H_1: \\mu < 49$$

$$t = \\frac{\\bar x - \\mu_0}{s/\\sqrt{n}} = \\frac{42 - 49}{3{,}4352} = -2{,}0377$$

Región de rechazo (cola izquierda, $\\alpha = 0{,}05$): $t < -t_{0.95,\\,11} = -1{,}7959$. El
estadístico cae dentro de la región de rechazo; el valor p es
$P(t_{11} < -2{,}0377) = 0{,}0332 < 0{,}05$. Se rechaza $H_0$: la evidencia muestral respalda
que el consumo promedio es menor a 49 $kWh$/año.

El mismo juicio se obtiene por dualidad con un intervalo unilatero: el límite superior con
confianza del 95% es

$$\\mu < \\bar x + t_{0.95,\\,11} \\frac{s}{\\sqrt{n}} = 42 + 1{,}7959 \\times 3{,}4352 = 48{,}17$$

Como $48{,}17 < 49$, se rechaza $H_0$ al 5%. Nótese la sutileza de la cola simple: el intervalo
bilateral del 95% **sí** contiene a 49 (su extremo superior es 49,56), pero la cola izquierda
al 5% usa el cuantil $t_{0.95}$ y no el $t_{0.975}$, y su umbral (48,17) es más conservador.
Niveles distintos, umbrales distintos.


```python

# --- EJ 1: aspiradora ZZ ---
n1, xbar1, s1 = 12, 42.0, 11.9
se1 = s1 / math.sqrt(n1)

print(f"se = s/sqrt(n) = {s1}/sqrt({n1}) = {se1:.6f}")

marg1 = Q["t_0.975_11"] * se1
ic1 = (xbar1 - marg1, xbar1 + marg1)
print(f"\na) IC 95%: {xbar1:.4f} +/- {Q['t_0.975_11']:.6f} x {se1:.6f} = {xbar1:.4f} +/- {marg1:.4f}")
print(f"   -> [{ic1[0]:.4f}, {ic1[1]:.4f}] kWh/anio   contiene 49: {ic1[0] <= 49 <= ic1[1]}")
print(f"   margen relativo: {marg1 / xbar1:.4f} del promedio")

mu0 = 49.0
t1 = (xbar1 - mu0) / se1
p1 = t_dist.cdf(t1, n1 - 1)
lim_sup1 = xbar1 + Q["t_0.95_11"] * se1
print(f"\nb) H0: mu = 49 vs H1: mu < 49   (alpha = 0.05)")
print(f"   t = ({xbar1:.1f} - 49)/{se1:.4f} = {t1:.4f}   vs   -t(0.95, 11) = {-Q['t_0.95_11']:.4f}")
print(f"   -> {'rechazar' if t1 < -Q['t_0.95_11'] else 'no rechazar'} H0")
print(f"   valor p = P(t11 < {t1:.4f}) = {p1:.4f}")
print(f"   limite superior unilatero 95%: {xbar1:.2f} + {Q['t_0.95_11']:.6f} x {se1:.4f} = {lim_sup1:.4f}")
print(f"   -> {'menor' if lim_sup1 < mu0 else 'mayor o igual'} a 49: la conclusion coincide con la prueba")

fig, axs = plt.subplots(1, 2, figsize=(9, 4.5))

x = np.linspace(-5, 5, 600)
y = t_dist.pdf(x, n1 - 1)
ax = axs[0]
ax.plot(x, y, color=AZUL, lw=2)
ax.fill_between(x, y, where=(x <= t1), color=ROJO, alpha=0.30)
ax.fill_between(x, y, where=(x >= -Q["t_0.95_11"]), color=GRIS, alpha=0.25)
ax.annotate(f"region de rechazo\n$\\alpha$ = 0,05", xy=(-3.2, t_dist.pdf(-3.2, 11)),
            xytext=(-4.9, 0.10), fontsize=10, color=ROJO,
            arrowprops=dict(arrowstyle="->", color=ROJO, lw=1))
ax.annotate(f"$t$ = {t1:.3f}", xy=(t1, t_dist.pdf(t1, 11)), xytext=(-1.9, 0.16),
            fontsize=10, color=AZUL,
            arrowprops=dict(arrowstyle="->", color=AZUL, lw=1))
ax.annotate(f"umbral $-t_{{0.95}}$ = {-Q['t_0.95_11']:.3f}",
            xy=(-Q["t_0.95_11"], t_dist.pdf(-Q["t_0.95_11"], 11)), xytext=(0.4, 0.24),
            fontsize=10, color=GRIS,
            arrowprops=dict(arrowstyle="->", color=GRIS, lw=1))
ax.set_xlabel("estadístico t")
ax.set_ylabel("densidad")
ax.set_ylim(0, 0.42)
ax.set_title(f"Prueba $H_1:\\ \\mu < 49$  (t de 11 gl)")

ax = axs[1]
ax.barh([0], [ic1[1] - ic1[0]], left=[ic1[0]], height=0.35, color=AZUL, alpha=0.35)
ax.plot([ic1[0], ic1[0]], [-0.2, 0.2], color=AZUL, lw=2.5)
ax.plot([ic1[1], ic1[1]], [-0.2, 0.2], color=AZUL, lw=2.5)
ax.plot([xbar1], [0], "o", color=AZUL, ms=7)
ax.plot([mu0], [0], "x", color=ROJO, ms=11, mew=2.5)
ax.annotate(f"IC 95%\n[{ic1[0]:.2f}; {ic1[1]:.2f}]", xy=((ic1[0] + ic1[1]) / 2, 0.22),
            ha="center", fontsize=10, color=AZUL)
ax.annotate(f"$\\mu_0$ = 49", xy=(mu0, -0.30), xytext=(mu0 + 0.4, -0.42),
            fontsize=10, color=ROJO)
ax.annotate(f"$\\bar x$ = {xbar1:.1f}", xy=(xbar1, -0.28), xytext=(xbar1 - 6.2, -0.42),
            fontsize=10, color=AZUL)
ax.set_xlim(30, 54)
ax.set_ylim(-0.5, 0.5)
ax.set_yticks([])
ax.grid(False, axis="y")
ax.set_xlabel("kWh por año")
ax.set_title("Intervalo de confianza del 95% para $\\mu$")

fig.tight_layout()
plt.show()
```

### Lectura de las figuras

**Izquierda:** la curva es la densidad de $t$ con 11 grados de libertad. La región gris de la
derecha es la de no rechazo y la roja de la izquierda la región de rechazo de $H_1:\\ \\mu < 49$
para $\\alpha = 0{,}05$, delimitada por el cuantil $-t_{0.95} = -1{,}796$. El estadístico observado
$t = -2{,}038$ cae del lado rojo: se rechaza. El valor p es el área roja **hasta** $t = -2{,}038$,
que es más chica que todo el $\\alpha$: $p = 0{,}033 < 0{,}05$.

**Derecha:** el intervalo $[34{,}44;\\ 49{,}56]$ contiene al $\\bar x = 42$ (círculo azul) y también
a $\\mu_0 = 49$ (cruz roja): por dualidad bilateral, esto indicaría que no se rechaza $H_0:\\ \\mu = 49$
con una prueba **de dos colas** al 5%. Pero la prueba del inciso es de **una cola** al 5%, cuyo
umbral es 48,17, y 49 queda afuera. No hay contradicción: son niveles de confianza distintos
(el intervalo bilateral del 90% es el que se alinea con la cola simple al 5%).


## 2. Ruido y rendimiento de una tarea: varianzas, diferencia de medias y cola simple

### Datos

Dos grupos independientes de 16 personas realizan la misma tarea; el primero con ruido normal
(menor a 40 dB) y el segundo con ruido alto (mayor a 50 dB). El rendimiento se mide en minutos.
Se suponen poblaciones normales con varianzas iguales y desconocidas.

| Magnitud | Ruido normal (1) | Ruido alto (2) |
|---|---|---|
| Tamaño muestral, $n$ | 16 | 16 |
| Promedio de la tarea, $\bar x$ (min) | 14,375 | 18,5 |
| Desviación estándar muestral, $s$ | 2,2767 | 2,4495 |
| Varianza muestral, $s^2$ | 5,1834 | 6,0001 |
| Grados de libertad, $n - 1$ | 15 | 15 |

### a) Prueba de igualdad de varianzas con $\\alpha = 0{,}10$

Pre-test previo a la t combinada:

$$H_0: \\sigma_1^2 = \\sigma_2^2 \\qquad H_1: \\sigma_1^2 \\ne \\sigma_2^2$$

$$F = \\frac{s_2^2}{s_1^2} = \\frac{6{,}0001}{5{,}1834} = 1{,}1576$$

con distribución $F_{15,\\,15}$ bajo la nula. Región de rechazo bilateral al 10%:
$F < F_{0.05,\\,15,\\,15} = 0{,}4161$ o $F > F_{0.95,\\,15,\\,15} = 2{,}4034$ (recíprocos entre sí,
por la simetría $F_{\\nu_1,\\nu_2} \\sim 1/F_{\\nu_2,\\nu_1}$ cuando $\\nu_1 = \\nu_2$). El estadístico
cae cómodamente en el intervalo de no rechazo; el valor p es $0{,}781$, muy superior a
$\\alpha = 0{,}10$. **No se rechaza** la igualdad de varianzas y procede la t combinada.

### b) Intervalo de confianza del 90% para la diferencia de promedios

Con el supuesto de varianzas iguales sostenido por el pre-test, la varianza combinada es

$$s_p^2 = \\frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2} =
\\frac{15 \\times 5{,}1834 + 15 \\times 6{,}0001}{30} = 5{,}5917 \\qquad s_p = 2{,}3647$$

$$\\text{error estándar} = s_p\\sqrt{\\frac{1}{n_1} + \\frac{1}{n_2}} = 2{,}3647 \\times
\\sqrt{\\frac{2}{16}} = 0{,}8360$$

$$\\text{IC}_{90\\%}: \\quad (\\bar x_1 - \\bar x_2) \\ \\pm \\ t_{0.95,\\,30} \\times 0{,}8360 =
-4{,}125 \\ \\pm \\ 1{,}419 = [\\,-5{,}544;\\ -2{,}706\\,]$$

con $\\nu = n_1 + n_2 - 2 = 30$ grados de libertad. El intervalo no contiene a cero y es
enteramente negativo: con confianza del 90%, el tiempo promedio con ruido alto excede al de
ruido normal entre 2,71 y 5,54 minutos. El ruido alto perjudica el rendimiento.

Supuestos enumerados: (i) normalidad de ambas poblaciones; (ii) independencia de las dos
muestras e independencia de las observaciones dentro de cada grupo; (iii) muestreo aleatorio;
(iv) varianzas poblacionales iguales ($\\sigma_1^2 = \\sigma_2^2 = \\sigma^2$), verificado en (a).

### c) Prueba de cola derecha con $\\alpha = 0{,}01$

¿Hay razón para creer que el tiempo promedio con ruido alto supera en **más de 2 minutos** al de
ruido normal? Con $\\delta = \\mu_2 - \\mu_1$:

$$H_0: \\delta \\le 2 \\qquad H_1: \\delta > 2$$

$$t = \\frac{(\\bar x_2 - \\bar x_1) - 2}{s_p\\sqrt{1/n_1 + 1/n_2}} = \\frac{(18{,}5 - 14{,}375) - 2}
{0{,}8360} = \\frac{2{,}125}{0{,}8360} = 2{,}5417$$

Región de rechazo (cola derecha, $\\alpha = 0{,}01$): $t > t_{0.99,\\,30} = 2{,}4573$. El
estadístico supera el umbral por escaso margen; el valor p es
$P(t_{30} > 2{,}5417) = 0{,}0082 < 0{,}01$. **Se rechaza $H_0$**: existe evidencia, al 1% de
significación, de que el exceso promedio de tiempo con ruido alto es mayor que 2 minutos.

Por dualidad, el límite inferior unilatero del 99% para $\\delta$ es

$$\\delta > (\\bar x_2 - \\bar x_1) - t_{0.99,\\,30} \\times 0{,}8360 = 4{,}125 - 2{,}054 = 2{,}071$$

y como $2{,}071 > 2$, la conclusión coincide.

## 3. NDMA en cerveza: disminución de más de 2 partes por billón

### Datos

Niveles de NDMA (sustancia carcinogénica, en partes por billón) en cerveza fabricada con el
proceso de secado anterior y con el proceso nuevo. Se suponen dos poblaciones normales
independientes con varianzas iguales.

| Magnitud | Proceso anterior (1) | Proceso nuevo (2) |
|---|---|---|
| Tamaño muestral, $n$ | 12 | 12 |
| Media muestral, $\bar x$ (ppb) | 5,25 | 1,5 |
| Varianza muestral, $s^2$ | 0,932 | 1,00 |
| Desviación estándar muestral, $s$ | 0,9654 | 1,0 |
| Grados de libertad, $n - 1$ | 11 | 11 |

### Prueba de hipótesis con $\\alpha = 0{,}05$

El enunciado fija de entrada el supuesto de varianzas iguales, así que se aplica directamente la
t combinada. Con $\\delta = \\mu_1 - \\mu_2$ (proceso anterior menos proceso nuevo), la hipótesis
de "disminución de **más de dos** partes por billón" es $\\delta > 2$:

$$H_0: \\delta \\le 2 \\qquad H_1: \\delta > 2$$

$$s_p^2 = \\frac{11 \\times 0{,}932 + 11 \\times 1{,}00}{22} = 0{,}9660 \\qquad s_p = 0{,}9829$$

$$\\text{error estándar} = s_p\\sqrt{\\frac{1}{12} + \\frac{1}{12}} = 0{,}9829 \\times
\\sqrt{\\frac{2}{12}} = 0{,}4012$$

$$t = \\frac{(\\bar x_1 - \\bar x_2) - 2}{s_p\\sqrt{1/n_1 + 1/n_2}} =
\\frac{(5{,}25 - 1{,}5) - 2}{0{,}4012} = \\frac{1{,}75}{0{,}4012} = 4{,}3614$$

con $\\nu = n_1 + n_2 - 2 = 22$ grados de libertad. Región de rechazo (cola derecha,
$\\alpha = 0{,}05$): $t > t_{0.95,\\,22} = 1{,}7171$. El estadístico supera el umbral por un
margen amplio; el valor p es $P(t_{22} > 4{,}3614) = 0{,}0001 < 0{,}05$. **Se rechaza $H_0$:**
hay evidencia, al 5% de significación, de que el proceso nuevo disminuyó el NDMA en más de dos
partes por billón en promedio.

Por dualidad, el límite inferior unilatero del 95% para la disminución promedio es

$$\\delta > (\\bar x_1 - \\bar x_2) - t_{0.95,\\,22} \\times 0{,}4012 = 3{,}75 - 0{,}689 = 3{,}061$$

Como $3{,}061 > 2$, la estimación por intervalos confirma el rechazo. A modo de control, el
intervalo bilateral del 95% para $\\delta$ es $3{,}75 \\pm 2{,}0739 \\times 0{,}4012 =
[\\,2{,}918;\\ 4{,}582\\,]$, que tampoco contiene a 2 (ni a 0): la caída de NDMA es grande y
estadísticamente clara, consistente con el objetivo del proceso nuevo.


```python

# --- EJ 3: NDMA ---
nA, xA, vA = 12, 5.25, 0.932     # proceso anterior
nB, xB, vB = 12, 1.5, 1.00       # proceso nuevo
gl3 = nA + nB - 2
sp3sq = ((nA - 1) * vA + (nB - 1) * vB) / gl3
sp3 = math.sqrt(sp3sq)
se3 = sp3 * math.sqrt(1 / nA + 1 / nB)
dif3 = xA - xB

print(f"sA^2 = {vA:.6f}   sB^2 = {vB:.6f}")
print(f"sp2 = (11 x {vA} + 11 x {vB})/{gl3} = {sp3sq:.4f},  sp = {sp3:.4f}")
print(f"se = {sp3:.4f} x sqrt(2/12) = {se3:.4f}")
print(f"dif = {xA} - {xB} = {dif3:.4f}")

t3 = (dif3 - 2.0) / se3
p3 = t_dist.sf(t3, gl3)
lim_inf3 = dif3 - Q["t_0.95_22"] * se3
ic_ctrl3 = (dif3 - Q["t_0.975_22"] * se3, dif3 + Q["t_0.975_22"] * se3)
print(f"\nH0: delta <= 2 vs H1: delta > 2   (alpha = 0.05)")
print(f"t = ({dif3:.4f} - 2)/{se3:.4f} = {t3:.4f}   vs   t(0.95, 22) = {Q['t_0.95_22']:.4f}")
print(f"-> {'rechazar' if t3 > Q['t_0.95_22'] else 'no rechazar'} H0;  valor p = P(t22 > {t3:.4f}) = {p3:.6f}")
print(f"limite inferior unilatero 95%: {dif3:.4f} - {Q['t_0.95_22']:.6f} x {se3:.4f} = {lim_inf3:.4f}  (> 2: {lim_inf3 > 2})")
print(f"control bilateral IC 95%: [{ic_ctrl3[0]:.4f}, {ic_ctrl3[1]:.4f}]   contiene 2: {ic_ctrl3[0] <= 2 <= ic_ctrl3[1]}")

fig, axs = plt.subplots(1, 2, figsize=(9, 4.5))

x = np.linspace(-4, 6.2, 700)
y = t_dist.pdf(x, gl3)
ax = axs[0]
ax.plot(x, y, color=AZUL, lw=2)
ax.fill_between(x, y, where=(x > Q["t_0.95_22"]), color=ROJO, alpha=0.30)
ax.fill_between(x, y, where=(x <= Q["t_0.95_22"]), color=GRIS, alpha=0.22)
ax.annotate(f"region de rechazo\n$\\alpha$ = 0,05", xy=(2.1, t_dist.pdf(2.1, gl3)),
            xytext=(2.35, 0.075), fontsize=10, color=ROJO,
            arrowprops=dict(arrowstyle="->", color=ROJO, lw=1))
ax.plot([t3], [0], "o", color=VERDE, ms=7)
ax.annotate(f"$t$ = {t3:.3f}\nvalor p = {p3:.4f}", xy=(t3, t_dist.pdf(t3, gl3)),
            xytext=(3.05, 0.13), fontsize=10, color=VERDE,
            arrowprops=dict(arrowstyle="->", color=VERDE, lw=1))
ax.annotate(f"$t_{{0.95}}$ = {Q['t_0.95_22']:.3f}", xy=(Q["t_0.95_22"], t_dist.pdf(Q["t_0.95_22"], gl3)),
            xytext=(1.0, 0.22), fontsize=10, color=GRIS)
ax.set_xlabel("estadístico t")
ax.set_ylabel("densidad")
ax.set_ylim(0, 0.42)
ax.set_title("Prueba $H_1:\\ \\delta > 2$  (t de 22 gl)")

ax = axs[1]
ax.barh([0], [ic_ctrl3[1] - ic_ctrl3[0]], left=[ic_ctrl3[0]], height=0.35, color=VERDE, alpha=0.30)
ax.plot([ic_ctrl3[0], ic_ctrl3[0]], [-0.2, 0.2], color=VERDE, lw=2.5)
ax.plot([ic_ctrl3[1], ic_ctrl3[1]], [-0.2, 0.2], color=VERDE, lw=2.5)
ax.plot([dif3], [0], "o", color=VERDE, ms=7)
ax.plot([2], [0], "x", color=ROJO, ms=11, mew=2.5)
ax.plot([lim_inf3, lim_inf3], [-0.2, 0.2], color=AZUL, lw=2, ls="--")
ax.annotate(f"IC 95% (control)\n[{ic_ctrl3[0]:.2f}; {ic_ctrl3[1]:.2f}]", xy=(dif3, 0.24),
            ha="center", fontsize=10, color=VERDE)
ax.annotate("$\\delta_0$ = 2", xy=(2, -0.28), xytext=(1.35, -0.44), fontsize=10, color=ROJO)
ax.annotate(f"limite unilatero 95%\n{lim_inf3:.3f}", xy=(lim_inf3, -0.20),
            xytext=(lim_inf3 - 0.15, -0.62), fontsize=10, color=AZUL, ha="center")
ax.annotate(f"$\\bar\\delta$ = {dif3:.2f}", xy=(dif3, -0.20), xytext=(dif3 + 0.42, -0.30),
            fontsize=10, color=VERDE)
ax.set_xlim(1.0, 5.1)
ax.set_ylim(-0.72, 0.6)
ax.set_yticks([])
ax.grid(False, axis="y")
ax.set_xlabel("partes por billón (anterior - nuevo)")
ax.set_title("Disminución promedio de NDMA: $\\delta = \\mu_1 - \\mu_2$")

fig.tight_layout()
plt.show()
```

## 4. Verificación por simulación

Los cuantiles tabulados sostienen decisiones enteras del desarrollo, así que conviene verificar
el comportamiento de los procedimientos por Monte Carlo. Con semilla fija
($\\texttt{default\\_rng(11)}$) y $M = 100\\,000$ réplicas, se simulan tres escenarios:

1. **Cobertura del intervalo del 95%** para $\\mu$ con una sola muestra normal de $n = 12$:
   se dibuja $M$ muestras de $N(0,1)$, se calcula $t = \\bar x/(s/\\sqrt{12})$ en cada una y se
   cuenta la fracción con $|t| \\le t_{0.975,\\,11} = 2{,}200985$. El valor teórico es 0,95.
2. **Tamaño de la prueba unilateral del 5%** de la sección 1: misma muestra, se cuenta la
   fracción con $t < -t_{0.95,\\,11} = -1{,}795885$ bajo la nula. El valor teórico es 0,05.
3. **Tamaño de la prueba combinada bilateral del 5%** con $n_1 = n_2 = 16$: dos muestras
   independientes de $N(0,1)$, varianza combinada como en la sección 2 y umbral
   $t_{0.975,\\,30} = 2{,}042272$. El valor teórico es 0,05.

La incertidumbre de cada proporción simulada es $\\sqrt{p(1-p)/M} \\approx 0{,}0007$; se compara
la desviación observada contra 3 desviaciones estándar ($\\approx 0{,}0021$).


```python

rng = np.random.default_rng(11)
M = 100_000

# 1) cobertura del IC 95% para mu con una muestra normal de n = 12
xa_mc = rng.normal(0.0, 1.0, size=(M, 12))
t_mc = xa_mc.mean(axis=1) / (xa_mc.std(axis=1, ddof=1) / math.sqrt(12))
cobertura = (np.abs(t_mc) <= Q["t_0.975_11"]).mean()

# 2) tamano de la prueba unilateral 5% con n = 12 (misma estructura de llamadas)
xb_mc = rng.normal(0.0, 1.0, size=(M, 12))
t_mc2 = xb_mc.mean(axis=1) / (xb_mc.std(axis=1, ddof=1) / math.sqrt(12))
alpha_uni = (t_mc2 < -Q["t_0.95_11"]).mean()

# 3) tamano de la prueba t combinada bilateral 5% con n1 = n2 = 16
y1 = rng.normal(0.0, 1.0, size=(M, 16))
y2 = rng.normal(0.0, 1.0, size=(M, 16))
m1, q1 = y1.mean(axis=1), y1.var(axis=1, ddof=1)
m2, q2 = y2.mean(axis=1), y2.var(axis=1, ddof=1)
sp2_mc = (15 * q1 + 15 * q2) / 30
t_mc3 = (m1 - m2) / (np.sqrt(sp2_mc) * math.sqrt(1 / 16 + 1 / 16))
alpha_pool = (np.abs(t_mc3) > Q["t_0.975_30"]).mean()

print(f"{'procedimiento':>28}{'teorico':>10}{'simulado':>10}{'desv':>10}{'3 sigmas':>10}   ok")
print("-" * 72)
resumen = []
for nom, teo, sim in [("cobertura IC 95% (n = 12)", 0.95, cobertura),
                      ("alpha unilateral 5% (n = 12)", 0.05, alpha_uni),
                      ("alpha combinada 5% (n = 16+16)", 0.05, alpha_pool)]:
    sig = math.sqrt(teo * (1 - teo) / M)
    dentro = abs(sim - teo) <= 3 * sig
    resumen.append(dentro)
    print(f"{nom:>28}{teo:>10.4f}{sim:>10.4f}{abs(sim - teo):>10.4f}{3 * sig:>10.4f}   {dentro}")
print()
print(f"semilla 11, M = {M:,}; las tres verificaciones caen dentro de 3 desviaciones: {all(resumen)}")

fig, axs = plt.subplots(1, 2, figsize=(9, 4.5))

etiquetas = ["cobertura\nIC 95%", "alpha\nunilateral 5%", "alpha\ncombinada 5%"]
teoricos = [0.95, 0.05, 0.05]
simulados = [cobertura, alpha_uni, alpha_pool]
pos = np.arange(3)

ax = axs[0]
ax.bar(pos - 0.2, teoricos, width=0.38, color=GRIS, alpha=0.65, label="teorico")
ax.bar(pos + 0.2, simulados, width=0.38, color=AZUL, alpha=0.85, label="simulado")
for i, (teo, sim) in enumerate(zip(teoricos, simulados)):
    ax.annotate(f"{teo:.3f}", xy=(i - 0.2, teo), xytext=(i - 0.2, teo + 0.03),
                ha="center", fontsize=9, color=GRIS)
    ax.annotate(f"{sim:.3f}", xy=(i + 0.2, sim), xytext=(i + 0.2, sim + 0.03),
                ha="center", fontsize=9, color=AZUL)
ax.set_xticks(pos)
ax.set_xticklabels(etiquetas)
ax.set_ylim(0, 1.08)
ax.set_ylabel("proporcion")
ax.legend(frameon=False, fontsize=10)
ax.set_title("Valor teorico vs simulado (M = 100.000)")

ax = axs[1]
difs = [(sim - teo) / math.sqrt(teo * (1 - teo) / M) for teo, sim in zip(teoricos, simulados)]
ax.axhline(0, color=GRIS, lw=1)
ax.axhline(3, color=ROJO, lw=1, ls="--")
ax.axhline(-3, color=ROJO, lw=1, ls="--")
ax.bar(pos, difs, width=0.45, color=VERDE, alpha=0.75)
for i, d in enumerate(difs):
    ax.annotate(f"{d:+.2f} sigma", xy=(i, d), xytext=(i, d + (0.25 if d >= 0 else -0.5)),
                ha="center", fontsize=10, color=VERDE)
ax.set_xticks(pos)
ax.set_xticklabels(etiquetas)
ax.set_ylim(-4.2, 4.2)
ax.set_ylabel("desviacion estandar")
ax.set_title("Desvio simulado-teorico en unidades de sigma")

fig.tight_layout()
plt.show()
```

### Lectura de las figuras

**Izquierda:** para los tres procedimientos, la barra gris (valor teórico) y la azul (valor
simulado con la semilla fija) son prácticamente indistinguibles: la cobertura del intervalo del
95% ronda 0,950 y los dos tamaños de prueba 0,05 (0,049 y 0,051). La simulación reproduce lo
que la teoría garantiza exactamente.

**Derecha:** los mismos desvíos medidos en desviaciones estándar de la proporción simulada.
Los tres quedan dentro de la banda $\\pm 3\\sigma$ (líneas rojas punteadas); el más extremo es el
tamaño de la prueba unilateral, a algo más de 1,3 sigma, dentro de lo esperable por azar. Los
cuantiles tabulados y la t combinada se comportan como corresponde en muestras de $n = 12$ y
$n = 16$ por grupo.


## Notas de validación

- Cuantiles $t$ y $F$ usados en el desarrollo contrastados contra `scipy.stats` (tabla de la
  sección marco teórico, con $|F(\\text{cuantil}) - q|$ evaluado por la CDF directa, no por su
  complemento). Se verifican además la simetría de $t$ ($t_{0.05,\\nu} = -t_{0.95,\\nu}$) y la
  reciprocidad $F_{0.05,\\,15,\\,15} = 1/F_{0.95,\\,15,\\,15}$.
- **Ejercicio 1:** el intervalo bilateral del 95%, $[34{,}44;\\ 49{,}56]$, contiene a $\\mu_0 = 49$,
  mientras que la prueba unilateral al 5% lo rechaza. No hay contradicción: la cola simple usa
  $t_{0.95} = 1{,}795885$ y su umbral es 48,1693, que es exactamente el extremo del intervalo
  bilateral del 90%. El valor p de la prueba es 0,0332.
- **Ejercicio 1:** con 11 grados de libertad, el cuantil $t$ agranda el margen un 12,3% respecto
  del que se obtendría con la normal ($7{,}561$ contra $6{,}733$ $kWh$): la cola pesada de $t$
  se paga en precisión.
- **Ejercicio 2:** el pre-test F da $p = 0{,}781$, sin motivo para cuestionar la igualdad de
  varianzas. Con tamaños iguales ($n_1 = n_2 = 16$) la t combinada es robusta ante varianzas
  desiguales: el control con Welch (gl $= 29{,}84$) da IC del 90% $[-5{,}5442;\\ -2{,}7058]$ frente
  a $[-5{,}5440;\\ -2{,}7060]$ de la t combinada, diferencia de cuarta decimal.
- **Ejercicio 2:** la decisión del inciso c es correcta pero justa: $t = 2{,}5417$ contra el
  umbral $2{,}4573$ ($p = 0{,}0082$). El IC del 90% del inciso b no sirve para decidir en c,
  porque es confianza bilateral del 90% (equivalente a cola simple del 95%) y c pide 1%.
- **Ejercicio 3:** el umbral de la hipótesis compuesta ($\\delta_0 = 2$) queda fuera también del
  intervalo bilateral de control $[2{,}918;\\ 4{,}582]$, y el límite del intervalo unilatero del
  95% es $3{,}061 > 2$: las dos vías (prueba e intervalo) coinciden sin sobresaltos.
- Monte Carlo con semilla 11 ($M = 100\\,000$): cobertura del IC del 95% y tamaños de las pruebas
  unilateral y combinada, todos dentro de 3 desviaciones estándar del valor teórico. El orden de
  los llamados a `rng.normal` está fijado en el código para que los resultados sean
  reproducibles.
- Los valores del marco teórico y del desarrollo están redondeados a 6 decimales en los cuantiles
  y a 4 en los estadísticos; las decisiones se toman siempre con el valor completo de
  `scipy.stats`.
