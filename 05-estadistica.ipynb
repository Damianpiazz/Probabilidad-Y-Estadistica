# Diferencia de dos medias y de dos proporciones: intervalos de confianza y prueba de hipótesis

Desarrollo de los ejercicios de comparación entre dos muestras independientes. Para la diferencia de
medias de poblaciones normales con varianzas conocidas y para la diferencia de proporciones se
construyen el estimador puntual, el intervalo de confianza y la prueba de hipótesis con su
estadístico, su distribución bajo la nula, la regla de decisión en escala estandarizada y en escala
de medición, y el valor p. Los resultados se contrastan entre sí (intervalo contra prueba) y por
simulación Monte Carlo con semilla fija.

## Marco teórico

### Dos muestras independientes

Sea $X_1, \dots, X_{n_1}$ una muestra de $N(\mu_1, \sigma_1^2)$ e $Y_1, \dots, Y_{n_2}$ una muestra
independiente de $N(\mu_2, \sigma_2^2)$, con las dos muestras mutuamente independientes. El
parámetro de interés es la diferencia $\delta = \mu_1 - \mu_2$ y el estimador puntual es la
diferencia de medias muestrales $\hat\delta = \bar X - \bar Y$. Por linealidad de la esperanza y la
varianza:

$$E[\bar X - \bar Y] = \mu_1 - \mu_2, \qquad \operatorname{Var}(\bar X - \bar Y) = \frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}$$

La varianza de la diferencia es la suma de las varianzas de cada media: así se comporta la
independencia, y por eso un diseño pareado no aplica acá (en muestras pareadas se trabaja con las
diferencias individuales y el error estándar es otro). Como ambas poblaciones son normales, la
diferencia de medias muestrales también lo es:

$$\hat\delta \ \dot\sim\ N\!\left(\mu_1 - \mu_2,\ \frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}\right), \qquad SE = \sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}$$

Cuando las varianzas poblacionales no se conocen se reemplazan por las varianzas muestrales; con
muestras grandes la sustitución desplaza apenas el error estándar y el cuantil normal sigue siendo
una buena aproximación.

### Intervalo de confianza para la diferencia de medias

$$(\bar X - \bar Y) \ \pm \ z_{\alpha/2} \sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}$$

El intervalo reúne los valores de $\mu_1 - \mu_2$ compatibles con las dos muestras. Si contiene al
cero, las muestras no muestran diferencia entre las medias al nivel del intervalo; si queda entero
de un lado, el signo de la diferencia está acotado. Que un intervalo contenga 0 no demuestra que las
medias sean iguales: indica que con ese tamaño de muestra la diferencia no se separa del ruido.

### Prueba de hipótesis para la diferencia de medias

$$H_0: \mu_1 - \mu_2 = \delta_0 \qquad \text{vs} \qquad H_1: \mu_1 - \mu_2 \ne \delta_0 \ \ (\text{o } > \delta_0,\ < \delta_0)$$

$$Z = \frac{(\bar X - \bar Y) - \delta_0}{\sqrt{\sigma_1^2/n_1 + \sigma_2^2/n_2}} \ \dot\sim\ N(0,1) \ \text{ bajo } H_0$$

La regla en escala estandarizada es rechazar si $z > z_\alpha$ (unilateral derecha) o si
$|z| > z_{\alpha/2}$ (bilateral); en escala de medición, rechazar si
$\hat\delta > \delta_0 + z_\alpha\, SE$. Los errores tipo I y tipo II se definen igual que para una
sola población, evaluando la distribución del estadístico bajo $H_0$ y bajo $H_1$. Para la prueba
unilateral derecha con diferencia real $\delta_1$:

$$\beta = \Phi\!\left(z_\alpha - \frac{|\delta_1 - \delta_0|}{SE}\right), \qquad \text{potencia} = 1 - \beta$$

La potencia depende de la diferencia real medida en errores estándar, de $\alpha$ y del tamaño de
muestra a través de $SE$, que decrece como $1/\sqrt{n}$. Con la diferencia fija, agrandar las
muestras es la única vía para subir la potencia.

### Diferencia de proporciones

Sean $X_1 \sim \text{Bin}(n_1, \pi_1)$ y $X_2 \sim \text{Bin}(n_2, \pi_2)$ independientes, con
$\hat p_i = X_i / n_i$. Entonces

$$E[\hat p_1 - \hat p_2] = \pi_1 - \pi_2, \qquad \operatorname{Var}(\hat p_1 - \hat p_2) = \frac{\pi_1(1-\pi_1)}{n_1} + \frac{\pi_2(1-\pi_2)}{n_2}$$

El intervalo de confianza usa la estimación muestral de cada varianza, sin agrupar nada:

$$(\hat p_1 - \hat p_2) \ \pm \ z_{\alpha/2} \sqrt{\frac{\hat p_1(1-\hat p_1)}{n_1} + \frac{\hat p_2(1-\hat p_2)}{n_2}}$$

### Prueba de hipótesis para la diferencia de proporciones

Bajo $H_0: \pi_1 = \pi_2$ las dos muestras provienen de una misma proporción $\pi$, desconocida. La
mejor estimación de ese $\pi$ combina los dos conteos (proporción agrupada):

$$\bar p = \frac{X_1 + X_2}{n_1 + n_2}, \qquad SE_{\text{pool}} = \sqrt{\bar p(1-\bar p)\left(\frac{1}{n_1} + \frac{1}{n_2}\right)}$$

$$Z = \frac{\hat p_1 - \hat p_2}{SE_{\text{pool}}} \ \dot\sim\ N(0,1) \ \text{ bajo } H_0$$

El agrupamiento es exclusivo de la prueba. El intervalo del inciso anterior no asume igualdad de
proporciones, por eso no agrupa; en cambio, bajo $H_0$ existe un único parámetro y usar las
varianzas muestrales por separado en el estadístico contradice la distribución derivada bajo la
nula. Condición de validez de la aproximación normal en cada grupo:
$n_i \hat p_i \ge 5$ y $n_i(1 - \hat p_i) \ge 5$.

### Dualidad entre intervalo y prueba

Un intervalo bilateral de confianza $1-\alpha$ y la prueba bilateral de nivel $\alpha$ dicen lo
mismo sobre $\delta_0$: si $\delta_0$ cae fuera del intervalo, la prueba lo rechaza, y viceversa. La
prueba unilateral se corresponde con el límite de un intervalo de una sola cola. Esa correspondencia
se usa como control cruzado entre los incisos del intervalo y de la prueba de cada ejercicio: los
dos tienen que coincidir en la conclusión.

### Cuantiles de referencia

Los cuantiles se eligen según la cola de rechazo, no según el nivel declarado del intervalo:

| Uso | Cola de rechazo $q$ | $z_q$ |
|---|---|---|
| Intervalo 95% o prueba bilateral 5% | 0.025 | 1.959964 |
| Intervalo 90% | 0.050 | 1.644854 |
| Prueba unilateral 5% | 0.050 | 1.644854 |
| Prueba unilateral 10% | 0.100 | 1.281552 |
| Prueba unilateral 1% | 0.010 | 2.326348 |


```python

import math

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

mpl.rcParams.update({
    "figure.figsize": (9, 4.5),
    "figure.dpi": 110,
    "axes.grid": True,
    "grid.alpha": 0.3,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
})

AZUL = "#1f4e79"
GRIS = "#808080"
VERDE = "#2e7d32"
ROJO = "#b03a2e"

# cuantiles de la normal estandar: clave = probabilidad de la cola superior P(Z > z)
Z = {0.01: 2.326348, 0.025: 1.959964, 0.05: 1.644854, 0.10: 1.281552}


def cola_normal(z):
    """P(Z > z) para Z ~ N(0,1), via la funcion error complementaria."""
    return 0.5 * math.erfc(z / math.sqrt(2))


def Phi(z):
    """P(Z <= z) para Z ~ N(0,1)."""
    return 0.5 * (1.0 + math.erf(z / math.sqrt(2)))


Phi_vec = np.vectorize(Phi)


print("numpy", np.__version__, "| matplotlib", mpl.__version__)
print()
print(f"{'cola q':>8}{'z_q':>12}{'P(Z > z_q)':>14}{'|diff|':>12}")
print("-" * 46)
for q in sorted(Z):
    z_q = Z[q]
    print(f"{q:>8.3f}{z_q:>12.6f}{cola_normal(z_q):>14.6f}{abs(cola_normal(z_q) - q):>12.1e}")
```

## 1. Consumo de pescado y nivel de colesterol: diferencia de medias

Un estudio tomó dos grupos de hombres: el primero sin consumo de pescado (ni un gramo por día) y el
segundo con consumo alto (más de 45 gramos por día). Veinte años después se registró el nivel de
colesterol en la dieta, uno de los factores de riesgo de enfermedad coronaria. Los datos:

| | Sin consumo de pescado | Consumo alto |
|---|---|---|
| Tamaño de muestra $n_i$ | 159 | 79 |
| Colesterol promedio $\bar x_i$ | 158 | 146 |
| Desviación estándar $s_i$ | 66 | 75 |

Los dos grupos se formaron por selección independiente (no son pares), de modo que aplica la teoría
de dos muestras independientes; las desviaciones estándar de la tabla se toman como las $\sigma_i$ de
la unidad, y con $n_1 = 159$ y $n_2 = 79$ la diferencia con usarlas como $s_i$ estimadas es
despreciable. La diferencia se define siempre como "sin consumo − consumo alto": valores positivos
indican más colesterol en el grupo que no come pescado.

### a) Intervalo de confianza del 90% para la diferencia

El error estándar de la diferencia es

$$SE = \sqrt{\frac{66^2}{159} + \frac{75^2}{79}} = \sqrt{27.3962 + 71.2025} = \sqrt{98.5988} = 9.9297$$

$$\hat\delta = 158 - 146 = 12, \qquad z_{0.05} = 1.644854$$

$$12 \pm 1.644854 \times 9.9297 = 12 \pm 16.3329 \ \Longrightarrow \ [-4.3329,\ 28.3329]$$

El intervalo contiene al cero: con estas muestras, la diferencia de niveles medios de colesterol es
compatible con valores desde 4.3 puntos a favor del grupo sin pescado hasta 28.3 puntos en contra.
No se detecta diferencia al 90%. Nótese que casi toda la anchura del intervalo viene del grupo de
consumo alto: con $n_2 = 79$ y $s_2 = 75$ su aporte a la varianza (71.20 de los 98.60 totales) casi
triplica el del grupo de 159 observaciones.

### b) Prueba de hipótesis al 5%

$$H_0: \mu_1 - \mu_2 = 0 \qquad \text{vs} \qquad H_1: \mu_1 - \mu_2 > 0$$

La alternativa es unilateral derecha: la hipótesis que hay que apoyar es que el consumo de pescado
reduce el colesterol, es decir que el grupo sin consumo queda por encima. La regla:

$$\text{rechazar } H_0 \ \text{ si } \ z > z_{0.05} = 1.644854 \quad (\text{escala de medición: } \hat\delta > 16.3329)$$

$$z = \frac{12}{9.9297} = 1.2085 \ < \ 1.644854, \qquad p = P(Z > 1.2085) = 0.113428$$

No se rechaza $H_0$: al 5% no hay evidencia de que el consumo de pescado reduzca el nivel de
colesterol. La conclusión es coherente con el inciso a): la prueba unilateral al 5% usa el mismo
cuantil (1.644854) que el intervalo bilateral del 90%, y el intervalo contenía al cero. El valor p
de 0.1134 lejos no está, pero la regla es la regla.

Conviene mirar también la potencia con la diferencia que se observó: si la diferencia real fuera de
12 puntos, esta prueba la detectaría apenas

$$1 - \beta = 1 - \Phi\!\left(1.644854 - \frac{12}{9.9297}\right) = 1 - \Phi(0.4364) = 0.3313$$

es decir, una de cada tres veces. Con estas muestras, no rechazar $H_0$ era el resultado más
probable aunque la hipótesis fuera cierta: el problema no es solo de la evidencia, es de tamaño
muestral frente al ruido (66 y 75 puntos de desviación estándar).

```python
n1, xbar1, s1 = 159, 158.0, 66.0
n2, xbar2, s2 = 79, 146.0, 75.0
alpha = 0.05

se1, se2 = s1 / math.sqrt(n1), s2 / math.sqrt(n2)
se = math.sqrt(s1**2 / n1 + s2**2 / n2)
dif = xbar1 - xbar2

z_ic = Z[0.05]
margen = z_ic * se
ic_lo, ic_hi = dif - margen, dif + margen

print(f"grupos: n1 = {n1}, xbar1 = {xbar1:.0f}, s1 = {s1:.0f}   |   n2 = {n2}, xbar2 = {xbar2:.0f}, s2 = {s2:.0f}")
print(f"error estandar de cada media: {s1:.0f}/sqrt({n1}) = {se1:.6f}   {s2:.0f}/sqrt({n2}) = {se2:.6f}")
print(f"varianza de la diferencia: {s1**2 / n1:.6f} + {s2**2 / n2:.6f} = {s1**2 / n1 + s2**2 / n2:.6f}")
print(f"error estandar de la diferencia: SE = {se:.6f}")
print(f"dif observada: {xbar1:.0f} - {xbar2:.0f} = {dif:.4f}")
print()
print(f"a) IC 90%: z_{{0.05}} = {z_ic:.6f}")
print(f"   {dif:.4f} +/- {z_ic:.6f} x {se:.6f} = {dif:.4f} +/- {margen:.4f} = [{ic_lo:.4f}, {ic_hi:.4f}]")
print(f"   contiene 0: {'si' if ic_lo <= 0 <= ic_hi else 'no'}")
print()

z_crit = Z[alpha]
z = dif / se
p_valor = cola_normal(z)
umbral = z_crit * se
print(f"b) H0: mu1 - mu2 = 0 vs H1: mu1 - mu2 > 0   |   alpha = {alpha}")
print(f"   regla estandarizada: rechazar si z > {z_crit:.6f}")
print(f"   regla en escala de medicion: dif > {z_crit:.6f} x {se:.6f} = {umbral:.4f}")
print(f"   z = {dif:.4f}/{se:.6f} = {z:.4f} -> {'rechazar H0' if z > z_crit else 'no rechazar H0'}")
print(f"   valor p = P(Z > {z:.4f}) = {p_valor:.6f}")
print()

ic1 = (xbar1 - z_ic * se1, xbar1 + z_ic * se1)
ic2 = (xbar2 - z_ic * se2, xbar2 + z_ic * se2)
print(f"IC 90% media sin consumo:  [{ic1[0]:.3f}, {ic1[1]:.3f}]")
print(f"IC 90% media consumo alto: [{ic2[0]:.3f}, {ic2[1]:.3f}]")

beta = Phi(z_crit - dif / se)
print(f"potencia con diferencia real {dif:.0f}: 1 - Phi({z_crit:.6f} - {dif:.4f}/{se:.6f}) = {1 - beta:.4f}")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(11, 4.2))

# panel izquierdo: IC 90% para cada media muestral
axs[0].hlines(1, ic1[0], ic1[1], color=AZUL, linewidth=3.4)
axs[0].plot(xbar1, 1, "o", color=AZUL, markersize=8, zorder=5)
axs[0].hlines(0, ic2[0], ic2[1], color=VERDE, linewidth=3.4)
axs[0].plot(xbar2, 0, "o", color=VERDE, markersize=8, zorder=5)
for y_pos, ic_par, color, media in [(1, ic1, AZUL, xbar1), (0, ic2, VERDE, xbar2)]:
    axs[0].text(ic_par[0] - 0.8, y_pos, f"{ic_par[0]:.1f}", ha="right", va="center",
                fontsize=9, color=color)
    axs[0].text(ic_par[1] + 0.8, y_pos, f"{ic_par[1]:.1f}", ha="left", va="center",
                fontsize=9, color=color)
    axs[0].text(media, y_pos - 0.34, f"{media:.0f}", ha="center", fontsize=9, color=color)
axs[0].set_yticks([0, 1], ["consumo alto (n = 79)", "sin consumo (n = 159)"])
axs[0].set_xlim(126, 174)
axs[0].set_ylim(-0.75, 1.75)
axs[0].set_xlabel("nivel de colesterol promedio")
axs[0].set_title("Intervalos de confianza del 90% para cada media")

# panel derecho: IC 90% para la diferencia
axs[1].hlines(0, ic_lo, ic_hi, color=AZUL, linewidth=4)
axs[1].plot(dif, 0, "o", color=AZUL, markersize=9, zorder=5)
axs[1].axvline(0, color=ROJO, linewidth=1.4, linestyle="--")
axs[1].text(-1.2, 0.34, "0: sin diferencia", color=ROJO, fontsize=9, ha="right")
axs[1].text(ic_lo - 1.0, 0, f"{ic_lo:.2f}", ha="right", va="center", fontsize=9, color=AZUL)
axs[1].text(ic_hi + 1.0, 0, f"{ic_hi:.2f}", ha="left", va="center", fontsize=9, color=AZUL)
axs[1].text(dif, 0.34, r"$\hat{\delta}$ = 12", ha="center", fontsize=9, color=AZUL)
axs[1].set_yticks([])
axs[1].set_xlim(-9, 37)
axs[1].set_ylim(-0.55, 0.62)
axs[1].set_xlabel(r"diferencia de medias (sin consumo $-$ consumo alto)")
axs[1].set_title("IC 90% para la diferencia: contiene al cero")

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(11, 4.2))

# panel izquierdo: escala de la diferencia (medicion)
malla = np.linspace(-24, 46, 1500)
dens = np.exp(-0.5 * (malla / se) ** 2) / (se * math.sqrt(2 * math.pi))
altura = dens.max()

axs[0].plot(malla, dens, color="black", linewidth=1.1)
axs[0].fill_between(malla, 0, dens, where=malla >= dif, color=GRIS, alpha=0.35,
                    label=f"valor p (p = {p_valor:.4f})")
axs[0].fill_between(malla, 0, dens, where=malla >= umbral, color=ROJO, alpha=0.6,
                    label="región de rechazo")
axs[0].axvline(umbral, color=ROJO, linewidth=1.3, linestyle="--")
axs[0].text(umbral + 1.2, altura * 0.40, f"umbral {umbral:.4f}", color=ROJO, fontsize=9)
axs[0].axvline(dif, color=AZUL, linewidth=1.5)
axs[0].text(dif + 1.2, altura * 0.72, r"$\hat{\delta}$ = 12", color=AZUL, fontsize=9)
axs[0].axvline(0, color=GRIS, linewidth=1.1, linestyle=":", label=r"$\delta_0$ = 0")
axs[0].set_xlim(-24, 46)
axs[0].set_ylim(0, altura * 1.2)
axs[0].set_xlabel(r"diferencia de medias $\bar{x}_1 - \bar{x}_2$")
axs[0].set_ylabel("densidad bajo $H_0$")
axs[0].set_title("Diferencia de medias bajo $H_0$: no se rechaza")
axs[0].legend(fontsize=9, loc="upper left")

# panel derecho: escala estandarizada
malla2 = np.linspace(-3.5, 4.5, 1000)
dens2 = np.exp(-0.5 * malla2**2) / math.sqrt(2 * math.pi)

axs[1].plot(malla2, dens2, color="black", linewidth=1.1)
axs[1].fill_between(malla2, 0, dens2, where=malla2 >= z_crit, color=ROJO, alpha=0.55,
                    label=r"región de rechazo ($\alpha$ = 0.05)")
axs[1].axvline(z_crit, color=ROJO, linewidth=1.3, linestyle="--")
axs[1].text(z_crit + 0.12, 0.06, r"$z_{0.05}$ = 1.6449", color=ROJO, fontsize=9,
            bbox=dict(facecolor="white", alpha=0.85, edgecolor="none", pad=1.5))
axs[1].axvline(z, color=AZUL, linewidth=1.5)
axs[1].annotate(f"z = {z:.4f}\n$p$ = {p_valor:.4f}", xy=(z, 0.19), xytext=(2.5, 0.30),
                color=AZUL, fontsize=9,
                bbox=dict(facecolor="white", alpha=0.85, edgecolor="none", pad=1.5),
                arrowprops=dict(arrowstyle="->", linewidth=0.9, color=AZUL))
axs[1].set_xlim(-3.5, 4.5)
axs[1].set_ylim(0, 0.46)
axs[1].set_xlabel(r"estadístico de prueba $z$")
axs[1].set_ylabel("densidad bajo $H_0$")
axs[1].set_title("Escala estandarizada: $z < z_{0.05}$")
axs[1].legend(fontsize=9, loc="upper left")

plt.tight_layout()
plt.show()
```

Las simulaciones siguientes repiten el procedimiento completo bajo cada régimen: con diferencia
verdadera 0 se mide la proporción de veces que la regla rechaza (el error tipo I) y cuántas veces el
intervalo construido contiene a la diferencia verdadera (la cobertura); con diferencia verdadera 12
se mide la potencia. Cobertura, error tipo I y potencia son tres caras de la misma regla, y las tres
tienen que cerrar contra sus valores teóricos con el único ruido del muestreo finito.

```python
rng = np.random.default_rng(11)
ensayos = 100_000

bajo_h0 = rng.normal(0.0, se, size=ensayos)
bajo_h1 = rng.normal(dif, se, size=ensayos)

alpha_emp = (bajo_h0 > umbral).mean()
cobertura_emp = (np.abs(bajo_h0) <= margen).mean()
potencia_emp = (bajo_h1 > umbral).mean()

potencia_calc = 1 - Phi(z_crit - dif / se)
nominales = [cobertura_emp, alpha_emp, potencia_emp]
teoricos = [0.90, alpha, potencia_calc]
err = [math.sqrt(t * (1 - t) / ensayos) for t in teoricos]

print(f"ensayos Monte Carlo: {ensayos:,} (semilla 11)")
print()
print(f"{'cantidad':>14}{'teorico':>12}{'simulado':>12}{'error muestral':>18}")
print("-" * 58)
for nombre, teo, sim, e in zip(["cobertura IC90", "alpha", "potencia"], teoricos, nominales, err):
    print(f"{nombre:>14}{teo:>12.4f}{sim:>12.4f}{e:>18.4f}")
print()
print(f"desviacion |simulado - teorico|: cobertura {abs(nominales[0] - teoricos[0]):.4f}"
      f"   alpha {abs(nominales[1] - teoricos[1]):.4f}"
      f"   potencia {abs(nominales[2] - teoricos[2]):.4f}")
print(f"limite de 3 sigmas: {3 * err[0]:.4f} / {3 * err[1]:.4f} / {3 * err[2]:.4f}")
```

## 2. Ingestión de vitamina A y cáncer pulmonar: diferencia de proporciones

Un estudio de 25 años investigó la protección que brinda la ingesta de vitamina A (caroteno) contra
el cáncer pulmonar. De 500 hombres con baja cantidad de esa sustancia, 15 desarrollaron cáncer
pulmonar; de 800 con consumo alto, solo 8.

| | Baja ingesta de vitamina A | Alta ingesta |
|---|---|---|
| Tamaño de muestra $n_i$ | 500 | 800 |
| Con cáncer pulmonar $x_i$ | 15 | 8 |
| Proporción muestral $\hat p_i$ | 0.030 | 0.010 |

La diferencia se define como "baja − alta": un valor positivo significa más cáncer en el grupo con
baja ingesta. Las condiciones de aproximación normal se cumplen en los cuatro conteos esperados:
15, 485, 8 y 792, todos mayores o iguales a 5.

### a) Intervalo de confianza del 95% para la diferencia de proporciones

$$SE = \sqrt{\frac{0.03 \times 0.97}{500} + \frac{0.01 \times 0.99}{800}} = \sqrt{0.00005820 + 0.00001238} = \sqrt{0.00007058} = 0.008401$$

$$\hat p_1 - \hat p_2 = 0.030 - 0.010 = 0.020, \qquad z_{0.025} = 1.959964$$

$$0.020 \pm 1.959964 \times 0.008401 = 0.020 \pm 0.016465 \ \Longrightarrow \ [0.003535,\ 0.036465]$$

El intervalo no contiene al cero y queda entero del lado positivo: la proporción de cáncer en el
grupo de baja ingesta es mayor, con una diferencia acotada entre 0.35 y 3.65 puntos porcentuales.
Al 95% ya se detecta la diferencia, con lo que cualquier prueba bilateral al 5% (y a fortiori la
unilateral al 10% del inciso b) rechazaría $H_0$.

### b) Prueba de hipótesis con $\alpha = 10\%$

$$H_0: \pi_1 - \pi_2 = 0 \qquad \text{vs} \qquad H_1: \pi_1 - \pi_2 > 0$$

La alternativa es unilateral derecha: lo que hay que demostrar es que la baja ingesta asocia más
cáncer (equivalente a que la alta ingesta protege). Bajo $H_0$ hay una sola proporción, estimada
con la proporción agrupada:

$$\bar p = \frac{15 + 8}{500 + 800} = \frac{23}{1300} = 0.0176923, \qquad SE_{\text{pool}} = \sqrt{0.0176923 \times 0.9823077 \times \left(\frac{1}{500} + \frac{1}{800}\right)} = 0.0075155$$

$$\text{rechazar } H_0 \ \text{ si } \ z > z_{0.10} = 1.281552$$

$$z = \frac{0.020}{0.0075155} = 2.6612 \ > \ 1.281552, \qquad p = P(Z > 2.6612) = 0.003894$$

Se rechaza $H_0$: al 10% (y también al 1%, por el margen) la evidencia indica que la proporción de
cáncer pulmonar es menor en el grupo de alta ingesta de vitamina A. El valor p de 0.0039 está muy
por debajo del nivel elegido: la conclusión no depende del 10%.

Dos precisiones sobre el método. Con el error estándar del intervalo (sin agrupar, 0.008401) el
estadístico daría $z = 2.3807$ con $p = 0.008640$, es decir también rechaza: la decisión no es
sensible a esa elección con estos datos, pero el cálculo bajo $H_0$ corresponde al agrupado. Y el
resultado es una asociación observacional en un estudio de cohorte, no una demostración causal: el
intervalo acota la diferencia de proporciones observadas entre los grupos.

```python
n1, x1 = 500, 15
n2, x2 = 800, 8
alpha = 0.10

p1, p2 = x1 / n1, x2 / n2
dif = p1 - p2

print(f"grupos: n1 = {n1}, x1 = {x1}, p1 = {p1:.4f}   |   n2 = {n2}, x2 = {x2}, p2 = {p2:.4f}")
print(f"condiciones de aproximacion: {x1}, {n1 - x1}, {x2}, {n2 - x2} -> todos >= 5:"
      f" {'si' if min(x1, n1 - x1, x2, n2 - x2) >= 5 else 'no'}")
print(f"dif observada: {p1:.3f} - {p2:.3f} = {dif:.4f}")
print()

z_ic = Z[0.025]
se_ic = math.sqrt(p1 * (1 - p1) / n1 + p2 * (1 - p2) / n2)
margen = z_ic * se_ic
ic_lo, ic_hi = dif - margen, dif + margen
print(f"a) IC 95%: z_{{0.025}} = {z_ic:.6f}")
print(f"   se por grupo: {p1:.3f} x {1 - p1:.3f}/{n1} = {p1 * (1 - p1) / n1:.8f}"
      f"   {p2:.3f} x {1 - p2:.3f}/{n2} = {p2 * (1 - p2) / n2:.8f}")
print(f"   SE (sin agrupar) = {se_ic:.6f}")
print(f"   {dif:.4f} +/- {z_ic:.6f} x {se_ic:.6f} = {dif:.4f} +/- {margen:.6f} = [{ic_lo:.6f}, {ic_hi:.6f}]")
print(f"   contiene 0: {'si' if ic_lo <= 0 <= ic_hi else 'no'}")
print()

z_crit = Z[alpha]
p_bar = (x1 + x2) / (n1 + n2)
se_pool = math.sqrt(p_bar * (1 - p_bar) * (1 / n1 + 1 / n2))
z = dif / se_pool
p_valor = cola_normal(z)
print(f"b) H0: pi1 - pi2 = 0 vs H1: pi1 - pi2 > 0   |   alpha = {alpha:.2f}")
print(f"   proporcion agrupada bajo H0: ({x1} + {x2})/({n1} + {n2}) = {p_bar:.7f}")
print(f"   SE_pool = {se_pool:.7f}")
print(f"   regla estandarizada: rechazar si z > {z_crit:.6f}")
print(f"   z = {dif:.4f}/{se_pool:.7f} = {z:.4f} -> {'rechazar H0' if z > z_crit else 'no rechazar H0'}")
print(f"   valor p = P(Z > {z:.4f}) = {p_valor:.6f}")
print()

z_sin = dif / se_ic
print(f"   control: con SE sin agrupar ({se_ic:.6f}): z = {z_sin:.4f}, p = {cola_normal(z_sin):.6f}")
print(f"            -> {'rechazar' if z_sin > z_crit else 'no rechazar'} H0 tambien")

delta_c = z_crit * se_pool
potencia_aprox = Phi((dif - delta_c) / se_ic)
print(f"   frontera en escala de la diferencia (SE congelado): {z_crit:.6f} x {se_pool:.7f} = {delta_c:.6f}")
print(f"   potencia aproximada con esa frontera: Phi(({dif:.4f} - {delta_c:.6f})/{se_ic:.6f})"
      f" = {potencia_aprox:.4f}")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(11, 4.2))

# panel izquierdo: IC 95% para cada proporción
se_a = math.sqrt(p1 * (1 - p1) / n1)
se_b = math.sqrt(p2 * (1 - p2) / n2)
ic_a = (p1 - z_ic * se_a, p1 + z_ic * se_a)
ic_b = (p2 - z_ic * se_b, p2 + z_ic * se_b)

axs[0].bar([0, 1], [p1, p2], width=0.46, color=[AZUL, VERDE], edgecolor="black", linewidth=0.7)
axs[0].errorbar([0, 1], [p1, p2],
                yerr=[[p1 - ic_a[0], p2 - ic_b[0]], [ic_a[1] - p1, ic_b[1] - p2]],
                fmt="none", ecolor="black", capsize=7, linewidth=1.1)
for x_pos, valor, ic_par, color in [(0, p1, ic_a, AZUL), (1, p2, ic_b, VERDE)]:
    axs[0].text(x_pos, ic_par[1] + 0.0018, f"{valor:.3f}", ha="center", fontsize=10, color=color)
    axs[0].text(x_pos, ic_par[1] + 0.0055,
                f"[{ic_par[0]:.4f}, {ic_par[1]:.4f}]", ha="center", fontsize=8.5)
axs[0].set_xticks([0, 1], ["baja ingesta\n(n = 500)", "alta ingesta\n(n = 800)"])
axs[0].set_xlim(-0.6, 1.6)
axs[0].set_ylim(0, 0.058)
axs[0].set_ylabel("proporción con cáncer pulmonar")
axs[0].set_title("Proporciones muestrales con IC 95%")

# panel derecho: IC para la diferencia de proporciones
malla = np.linspace(-0.010, 0.052, 1200)
dens = np.exp(-0.5 * ((malla - dif) / se_ic) ** 2) / (se_ic * math.sqrt(2 * math.pi))
altura = dens.max()

axs[1].plot(malla, dens, color="black", linewidth=1.1)
axs[1].fill_between(malla, 0, dens, where=(malla >= ic_lo) & (malla <= ic_hi), color=AZUL,
                    alpha=0.35, label="IC 95% (0.95 de cobertura)")
axs[1].axvline(ic_lo, color=AZUL, linewidth=1.2, linestyle="--")
axs[1].axvline(ic_hi, color=AZUL, linewidth=1.2, linestyle="--")
axs[1].axvline(dif, color=AZUL, linewidth=1.5)
axs[1].axvline(0, color=ROJO, linewidth=1.6, linestyle=":")
axs[1].text(0.0012, altura * 0.70, "0: fuera\ndel intervalo", color=ROJO, fontsize=9)
axs[1].text(dif, altura * 1.04, r"$\hat{p}_1 - \hat{p}_2$ = 0.020", ha="center",
            fontsize=9, color=AZUL)
axs[1].text(ic_lo, altura * 0.30, f"{ic_lo:.6f}", ha="right", fontsize=8.5, color=AZUL,
            rotation=90, va="center",
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none", pad=1))
axs[1].text(ic_hi, altura * 0.30, f"{ic_hi:.6f}", ha="left", fontsize=8.5, color=AZUL,
            rotation=90, va="center",
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none", pad=1))
axs[1].set_xlim(-0.010, 0.052)
axs[1].set_ylim(0, altura * 1.22)
axs[1].set_xlabel(r"diferencia de proporciones (baja $-$ alta)")
axs[1].set_ylabel("densidad de $\\hat{p}_1 - \\hat{p}_2$")
axs[1].set_title("IC 95% para la diferencia de proporciones")
axs[1].legend(fontsize=9, loc="upper right")

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(11, 4.2))

# panel izquierdo: escala de la diferencia con las distribuciones bajo H0 y H1
malla = np.linspace(-0.030, 0.058, 1400)
dens0 = np.exp(-0.5 * (malla / se_pool) ** 2) / (se_pool * math.sqrt(2 * math.pi))
dens1 = np.exp(-0.5 * ((malla - dif) / se_ic) ** 2) / (se_ic * math.sqrt(2 * math.pi))

axs[0].plot(malla, dens0, color="black", linewidth=1.1, label=r"$H_0$: diferencia = 0")
axs[0].fill_between(malla, 0, dens0, where=malla >= delta_c, color=ROJO, alpha=0.55,
                    label=r"región de rechazo ($\alpha$ = 0.10)")
axs[0].plot(malla, dens1, color=AZUL, linewidth=1.4, label=r"$H_1$: diferencia = 0.02")
axs[0].fill_between(malla, 0, dens1, where=malla >= delta_c, color=AZUL, alpha=0.30,
                    label="potencia (aprox.)")
axs[0].axvline(delta_c, color="black", linewidth=1.3, linestyle="--")
axs[0].text(delta_c + 0.0012, 60, f"frontera {delta_c:.6f}", fontsize=9,
            bbox=dict(facecolor="white", alpha=0.85, edgecolor="none", pad=1.5))
axs[0].axvline(dif, color=VERDE, linewidth=1.5)
axs[0].text(dif + 0.0012, 30, r"$\hat{\delta}$ = 0.020", fontsize=9, color=VERDE,
            bbox=dict(facecolor="white", alpha=0.85, edgecolor="none", pad=1.5))
axs[0].set_xlim(-0.030, 0.058)
axs[0].set_ylim(0, 82)
axs[0].set_xlabel(r"diferencia de proporciones (baja $-$ alta)")
axs[0].set_ylabel("densidad de la diferencia muestral")
axs[0].set_title("Distribución de la diferencia bajo $H_0$ y bajo $H_1$")
axs[0].legend(fontsize=9, loc="upper left")

# panel derecho: escala estandarizada
malla2 = np.linspace(-4, 5, 1000)
dens2 = np.exp(-0.5 * malla2**2) / math.sqrt(2 * math.pi)

axs[1].plot(malla2, dens2, color="black", linewidth=1.1)
axs[1].fill_between(malla2, 0, dens2, where=malla2 >= z_crit, color=ROJO, alpha=0.55,
                    label=r"región de rechazo ($\alpha$ = 0.10)")
axs[1].axvline(z_crit, color=ROJO, linewidth=1.3, linestyle="--")
axs[1].text(z_crit + 0.12, 0.07, r"$z_{0.10}$ = 1.2816", color=ROJO, fontsize=9,
            bbox=dict(facecolor="white", alpha=0.85, edgecolor="none", pad=1.5))
axs[1].axvline(z, color=AZUL, linewidth=1.5)
axs[1].annotate(f"z = {z:.4f}\n$p$ = {p_valor:.4f}", xy=(z, 0.10), xytext=(3.2, 0.26),
                color=AZUL, fontsize=9,
                bbox=dict(facecolor="white", alpha=0.85, edgecolor="none", pad=1.5),
                arrowprops=dict(arrowstyle="->", linewidth=0.9, color=AZUL))
axs[1].set_xlim(-4, 5)
axs[1].set_ylim(0, 0.44)
axs[1].set_xlabel(r"estadístico de prueba $z$")
axs[1].set_ylabel("densidad bajo $H_0$")
axs[1].set_title("Escala estandarizada: $z > z_{0.10}$, se rechaza")
axs[1].legend(fontsize=9, loc="upper left")

plt.tight_layout()
plt.show()
```

La aproximación normal deja reparos con proporciones tan bajas como 0.03 y 0.01, y conviene
medirlos antes que suponerlos. El intervalo de Wald (la fórmula con varianzas muestrales) no llega a
la cobertura nominal con eventos raros, y el error estándar agrupado de la prueba tampoco es un
número fijo: se recalcula en cada muestra. Ambos efectos se computan exactamente recorriendo la
grilla completa de pares $(x_1, x_2)$ —501 por 801— y ponderando cada par con las masas binomiales
de su grupo. La suma sobre la grilla es la probabilidad exacta de cubrir, de rechazar bajo $H_0$ y
de rechazar bajo $H_1$, sin simulación ni aproximación.

```python
def binom_pmf(n_, p_):
    """Masa Binomial(n_, p_) en la grilla 0..n_, via logaritmos para evitar desborde."""
    k = np.arange(n_ + 1)
    log_c = np.array([math.lgamma(n_ + 1) - math.lgamma(int(i) + 1) - math.lgamma(n_ - int(i) + 1)
                      for i in k])
    return np.exp(log_c + k * math.log(p_) + (n_ - k) * math.log(1 - p_))


pmf1, pmf2 = binom_pmf(n1, p1), binom_pmf(n2, p2)
pmf1_h0, pmf2_h0 = binom_pmf(n1, p_bar), binom_pmf(n2, p_bar)

X1 = np.arange(n1 + 1)[:, None]
X2 = np.arange(n2 + 1)[None, :]
p_a, p_b = X1 / n1, X2 / n2
pesos_h1 = pmf1[:, None] * pmf2[None, :]
pesos_h0 = pmf1_h0[:, None] * pmf2_h0[None, :]

# cobertura exacta del IC 95% con la diferencia verdadera 0.02
se_grilla = np.sqrt(p_a * (1 - p_a) / n1 + p_b * (1 - p_b) / n2)
masc_cobertura = np.abs((p_a - p_b) - dif) <= z_ic * se_grilla
cobertura_exacta = float((masc_cobertura * pesos_h1).sum())

# tamano exacto de la prueba bajo H0 con la proporcion comun agrupada
se_h0 = math.sqrt(p_bar * (1 - p_bar) * (1 / n1 + 1 / n2))
masc_rechazo = (p_a - p_b) / se_h0 > z_crit
tamano_exacto = float((masc_rechazo * pesos_h0).sum())

# potencia exacta: el estadistico recalcula su SE agrupado en cada par de conteos
p_mix = (X1 + X2) / (n1 + n2)
with np.errstate(divide="ignore", invalid="ignore"):
    z_grilla = (p_a - p_b) / np.sqrt(p_mix * (1 - p_mix) * (1 / n1 + 1 / n2))
masc_potencia = np.where(np.isfinite(z_grilla), z_grilla > z_crit, False)
potencia_exacta = float((masc_potencia * pesos_h1).sum())

print(f"control de la grilla: sumas de las masas {pmf1.sum():.10f} / {pmf2.sum():.10f}"
      f"   pares {pesos_h1.size:,}")
print()
print(f"{'cantidad':>14}{'normal':>10}{'exacto':>12}{'dif.':>10}")
print("-" * 46)
print(f"{'cobertura IC95':>14}{0.95:>10.4f}{cobertura_exacta:>12.6f}{cobertura_exacta - 0.95:>10.4f}")
print(f"{'alpha':>14}{alpha:>10.4f}{tamano_exacto:>12.6f}{tamano_exacto - alpha:>10.4f}")
print(f"{'potencia':>14}{potencia_aprox:>10.4f}{potencia_exacta:>12.6f}"
      f"{potencia_exacta - potencia_aprox:>10.4f}")
```

Las simulaciones repiten ahora el procedimiento con muestreo aleatorio. Bajo $H_0$ se fija la
proporción común en la agrupada observada (0.0176923) y se cuenta cuántas veces el estadístico
rechaza; con las proporciones 0.03 y 0.01 se miden la potencia y la cobertura del intervalo del
95%. La comparación se hace contra los valores exactos de la grilla, que son los que la simulación
debe reproducir; la columna de aproximación normal queda como referencia de lo que da el cálculo
analítico sin corrección.

```python
rng = np.random.default_rng(11)
ensayos = 100_000


def z_pooled(x_a, x_b):
    """Estadistico de la diferencia de proporciones con error estandar agrupado."""
    p_a, p_b = x_a / n1, x_b / n2
    p_mix = (x_a + x_b) / (n1 + n2)
    se_mix = np.sqrt(p_mix * (1 - p_mix) * (1 / n1 + 1 / n2))
    return (p_a - p_b) / se_mix


# bajo H0: una sola proporcion, la agrupada observada
x0_a = rng.binomial(n1, p_bar, size=ensayos)
x0_b = rng.binomial(n2, p_bar, size=ensayos)
z0 = z_pooled(x0_a, x0_b)
alpha_emp = (z0 > z_crit).mean()

# bajo H1: proporciones 0.03 y 0.01
x1_a = rng.binomial(n1, p1, size=ensayos)
x1_b = rng.binomial(n2, p2, size=ensayos)
z1 = z_pooled(x1_a, x1_b)
potencia_emp = (z1 > z_crit).mean()

# cobertura del IC 95% con la diferencia verdadera igual a la observada
p1_sim, p2_sim = x1_a / n1, x1_b / n2
se_sim = np.sqrt(p1_sim * (1 - p1_sim) / n1 + p2_sim * (1 - p2_sim) / n2)
cobertura_emp = (np.abs((p1_sim - p2_sim) - dif) <= z_ic * se_sim).mean()

normales = [0.95, alpha, potencia_aprox]
exactos = [cobertura_exacta, tamano_exacto, potencia_exacta]
simulados = [cobertura_emp, alpha_emp, potencia_emp]
err = [math.sqrt(t * (1 - t) / ensayos) for t in exactos]

print(f"ensayos Monte Carlo: {ensayos:,} (semilla 11), prop comun bajo H0 = {p_bar:.7f}")
print()
print(f"{'cantidad':>14}{'normal':>10}{'exacto':>10}{'simulado':>10}{'3 sigmas':>12}")
print("-" * 56)
for nombre, nor, exa, sim, e in zip(["cobertura IC95", "alpha", "potencia"],
                                    normales, exactos, simulados, err):
    print(f"{nombre:>14}{nor:>10.4f}{exa:>10.4f}{sim:>10.4f}{3 * e:>12.4f}")
print()
print(f"desviacion |simulado - exacto|: cobertura {abs(simulados[0] - exactos[0]):.4f}"
      f"   alpha {abs(simulados[1] - exactos[1]):.4f}"
      f"   potencia {abs(simulados[2] - exactos[2]):.4f}")
print(f"limite de 3 sigmas: {3 * err[0]:.4f} / {3 * err[1]:.4f} / {3 * err[2]:.4f}")
```

## Notas de validación

- Los cuatro cuantiles de la tabla del marco teórico se verifican con la función error
  complementaria: las colas resultantes son 0.010000, 0.025000, 0.050000 y 0.100000, exactamente las
  probabilidades de cola para las que están tabulados. La clave del diccionario es la probabilidad de
  la cola de rechazo, no el nivel del intervalo: el 90% de confianza usa el cuantil de cola 0.05.
- Resumen de resultados: sección 1, $SE = 9.9297$, IC 90% $[-4.3329, 28.3329]$ (contiene 0),
  $z = 1.2085 < 1.644854$, $p = 0.113428$, no se rechaza, con potencia 0.3313 si la diferencia real
  fuera de 12 puntos; sección 2, $SE = 0.008401$, IC 95% $[0.003535, 0.036465]$ (no contiene 0),
  $\bar p = 0.0176923$, $SE_{\text{pool}} = 0.0075155$, $z = 2.6612 > 1.281552$, $p = 0.003894$, se
  rechaza.
- El control cruzado entre incisos cierra en los dos ejercicios: en la sección 1, el IC del 90%
  contiene al 0 y la prueba unilateral al 5% (mismo cuantil 1.644854) no rechaza; en la sección 2, el
  IC del 95% no contiene al 0 y la prueba unilateral al 10% rechaza, con un valor p (0.0039) que
  rechazaría hasta al 1%. Intervalo y prueba son la misma decisión contada de dos formas, y acá
  coinciden en los dos casos.
- En la sección 1 la varianza de la diferencia está dominada por el grupo menor:
  $5625/79 = 71.20$ contra $4356/159 = 27.40$, es decir, el grupo de 79 observaciones aporta el 72%
  de la varianza del error estándar. Es el motivo de que el IC mida 32.7 puntos con una diferencia
  observada de 12: para demostrar el efecto haría falta o bien una diferencia mayor o bien más
  observaciones en el grupo de consumo alto.
- Con las desviaciones estándar de las muestras (66 y 75) en lugar de las poblacionales, un t de
  Welch con unos 140 grados de libertad daría $t_{0.05} \approx 1.656$ en vez de 1.644854 y el
  intervalo se ensancharía unos 0.11 puntos; la diferencia no cambia ninguna conclusión. Se mantiene
  el cuantil normal por ser el planteo de la unidad (varianzas conocidas) y por el tamaño de las
  muestras.
- En la sección 2 el agrupamiento de la proporción se aplica solo en la prueba, porque es ahí donde
  $H_0$ impone una única proporción. Como control, con el error estándar sin agrupar el estadístico
  daría $z = 2.3807$ con $p = 0.008640$: también rechaza, de modo que la conclusión no depende de
  esa elección con estos datos, aunque el cálculo que corresponde bajo la nula es el agrupado.
- La grilla binomial de la sección 2 (501 x 801 pares, con masas por logaritmo para evitar
  desborde) da los valores exactos: cobertura del IC de Wald 0.944033, tamaño de la prueba bajo $H_0$
  0.101196 y potencia 0.903886. Los tres difieren de la aproximación normal (0.95, 0.10 y 0.8914)
  por motivos distintos y previsibles: la cobertura de Wald queda por debajo del nominal con
  eventos raros (solo 15 y 8 cánceres), el tamaño crece un poco porque la cola discreta del
  estadístico cae distinto que la normal, y la potencia sube porque el error estándar agrupado no es
  constante sino que se recalcula muestra a muestra (su desviación bajo $H_1$ es 0.0036 contra un
  valor de 0.0177). Ninguno de los tres es ruido: son el sesgo del método, y por eso la simulación
  se contrasta contra el exacto y no contra el analítico.
- Las simulaciones Monte Carlo de ambas secciones usan 100.000 ensayos con semilla 11. En la
  sección 1 (modelo normal, donde la teoría es exacta) simulado y teórico quedan dentro de tres
  errores muestrales $\sqrt{p(1-p)/M}$: cobertura 0.9000, $\alpha$ 0.0493 y potencia 0.3272 contra
  0.9000, 0.0500 y 0.3313, con desviaciones de 0.0000, 0.0007 y 0.0040 frente a límites de 0.0028,
  0.0021 y 0.0045. En la sección 2 el mismo criterio se cumple contra los valores exactos de la
  grilla: desviaciones 0.0011, 0.0024 y 0.0019 frente a límites de 0.0022, 0.0029 y 0.0028.
- No rechazar $H_0$ en la sección 1 no es aceptar que las medias son iguales: el IC del 90% admite
  diferencias de hasta 28 puntos a favor del grupo sin consumo. La potencia 0.3313 explica el
  resultado: con ese tamaño de muestra, no rechazar era la salida más probable incluso si el pescado
  ayudara exactamente en la magnitud observada.
